```python
COUNTRY = "Sudan"
FILE = "../data/sudan.csv"
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import zscore

plt.style.use("default")
```

```python
df = pd.read_csv(FILE)

df["Country"] = COUNTRY
df.head()
```

```python
df["DATE"] = pd.to_datetime(df["YEAR"] * 1000 + df["DOY"], format="%Y%j")
df["Month"] = df["DATE"].dt.month
```

```python
df.replace(-999, np.nan, inplace=True)

missing_percent = df.isna().mean() * 100
missing_percent.sort_values(ascending=False)
```

```python
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df.describe()
```

Interpret key patterns in temperature, rainfall, and humidity.

```python
cols = ["T2M", "T2M_MAX", "T2M_MIN", "PRECTOTCORR", "RH2M", "WS2M"]

z = np.abs(zscore(df[cols], nan_policy="omit"))

outliers = (z > 3).sum()
outliers
```

```python
monthly_temp = df.resample("M", on="DATE")["T2M"].mean()

plt.figure()
monthly_temp.plot()
plt.title(f"{COUNTRY} Monthly Temperature Trend")
plt.ylabel("°C")
plt.show()
```

```python
monthly_rain = df.resample("M", on="DATE")["PRECTOTCORR"].sum()

plt.figure()
monthly_rain.plot()
plt.title(f"{COUNTRY} Monthly Rainfall Trend")
plt.ylabel("mm/day")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=False, cmap="coolwarm")
plt.title(f"{COUNTRY} Correlation Matrix")
plt.show()
```

```python
plt.figure()
df["PRECTOTCORR"].hist(bins=50)
plt.title(f"{COUNTRY} Rainfall Distribution")
plt.show()
```

```python
plt.figure()
plt.scatter(df["T2M"], df["RH2M"], alpha=0.3)
plt.title(f"{COUNTRY} Temperature vs Humidity")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.show()
```

## Key Insights - NIGERIA

- Temperature shows an overall upward trend.
- Rainfall is highly variable with seasonal spikes.
- Heat and humidity show negative correlation, indicating drying conditions.
- Extreme values suggest real climate stress events.
